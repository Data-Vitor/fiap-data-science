Probabilidade - arranjos simples em python

```python
from itertools import permutations

#lista de 5 livros
livros = ['A','B','C','D','E']

#Gerar todos os arranjos simples de 3 livros
arranjos = list(permutations(livros,3))
print(f'Numero de arranjos possiveis: {len(arranjos)} ')
print('Exemplos de arranjos: ')
for arranjo in arranjos[:5]:
    print(arranjo)
```

Probabilidade - Arranjos com repetição em python

```python
from itertools import product
digitos = [str(i) for i in range(1,7)]
senhas = [''.join(s) for s in product(digitos, repeat =4)]

print('Total de senhas possiveis', len(senhas))

```

probabilidade - combinaçãos simples em **python** - utilizada para saber de quantas maneiras é possivel selecionar elementos de um conjunto, **SEM SE IMPORTAR COM A ORDEM** e **sem repetir elementos**

```python
from itertools import combinations

pessoas = ['A', 'B', 'C', 'D', 'E' ]
combinacoes = list(combinations(pessoas,3))
print('numero de combinações possiveis:', len(combinacoes))

for c in combinacoes[:5]:
  print(c)

```

Probabilidade - combinaçãos com repetição em **Python** - utilizada para saber de quantas maneiras é possivel selecionar **REPETIDOS** de de um conjunta **não importa a ordem**

```python
from itertools import combinations_with_replacement

sabores = ['morango', 'chocolate','creme']
comb_repet = list(combinations_with_replacement(sabores,2))

print('Numero de combinações com repetição:',len(comb_repet))
print('Exemplos de combinações')
for c in comb_repet:
  print(c)
```

**Probabilidade-Amostragem** - Cálculo consolidado em **Python** do tamanho mínimo de amostra para quatro cenários: população infitia e finita, Media em população infinita e finita

```python
from math import ceil
#parametros gerais

z= 1.96 #z-score de 05%de confiança, obtido  da tabela normal
p = 0.5 #proporção esperada (caso conservador)
E_prop = 0.05 #margem de erro para proporção
S = 4 #desvio padrão estimado (var continua)
E_mean = 1 #erro máximo tolerado para a media

#populçao finitas
n_prop = 800
n_mean = 500

#1. proporção em populacao infinita
n0_prop_inf = (z**2 * p * (1-p)) / (E_prop**2)
print('Proporção - população infinita:', ceil(n0_prop_inf))

#2. Proporção em população finita
n_prop_finita = n0_prop_inf / (1+ (n0_prop_inf-1)/ n_prop)
print('Proporção de população finita:', ceil(n_prop_finita))

#3. Media em populçao infinita
n0_mean_inf = (z**2 * S**2) / (E_mean**2)
print('Média - populçaõ infinta:', ceil(n0_mean_inf))

#4. Média em população finita
n_mean_finita = n0_mean_inf / (1+(n0_mean_inf-1)/ n_mean)
print('Média - população finita:', ceil(n_mean_finita))
```

**3.** A população de uma ilha foi atacada por uma nova doença. Entre os habitantes, a probabilidade de estar doente é de 1%. Um método de detecção dessa doença foi desenvolvido e, quando a pessoa está doente, ele tem 99% de precisão (sensibilidade). Para as pessoas que não estão doentes, o método tem 95% de precisão (especificidade). Sabendo disso, qual a chance de estar realmente doente se o teste der positivo? Use o Teorema de Bayes para resolver este problema.

```python
from math import ceil
#resolvendo do jeito mais intuitivo com um número de população de exemplo

sensibilidade = 0.99
especificidade = 0.95

populacao =10000
prop_doente = 0.01
prop_nao_doente = 0.95
qtd_doentes = populacao * prop_doente
qtd_saudaveis = populacao - qtd_doentes

#primeiro grupo
_doente = qtd_doentes * sensibilidade
falso_positivo_doente = qtd_doentes - _doente



#segundo bloco
realmente_saudavel = qtd_saudaveis*prop_nao_doente
falso_positivo_saudavel = qtd_saudaveis -realmente_saudavel


#qual a chance de realmente estar doente se o teste for positivo?

realmente_doente = (_doente) / (_doente+falso_positivo_saudavel)


print(f'A chance de realmente estar doente se o teste der positivo é de aproximadamente :{ceil(realmente_doente*100)}%')


```
